```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])

print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

### Tugas Analisis 1

**Pertanyaan:**
Bandingkan hasil `harga * 0.9` di atas dengan jika harga berupa list Python biasa (bukan array). Apa yang terjadi bila kamu mencoba `[5000,7000,3000] * 0.9`? Jelaskan mengapa berbeda.

**Jawaban:**
Pada NumPy array, `harga * 0.9` dapat menghitung harga setelah diskon 10% untuk semua data sekaligus. Hasilnya adalah `[4500., 6300., 2700., 10800., 4050.]`. Sedangkan pada list Python biasa, `[5000,7000,3000] * 0.9` menghasilkan error karena list Python tidak dapat dikalikan langsung dengan bilangan desimal. Perbedaannya karena NumPy array mendukung operasi matematika secara langsung pada setiap elemen.


```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)

print(df)
```

### Tugas Analisis 2

**Pertanyaan:**
Amati tabel yang dihasilkan. Kolom mana yang terlihat memiliki data kosong (None)? Menurutmu apa risikonya jika data ini langsung dianalisis tanpa dibersihkan?

**Jawaban:**
Berdasarkan data pada Latihan 2, terdapat data kosong pada kolom `menu` dan `terjual`. Kolom `menu` memiliki satu data `None`, sedangkan kolom `terjual` juga memiliki satu data kosong. Jika data langsung dianalisis tanpa dibersihkan, hasil analisis dapat menjadi kurang tepat karena terdapat informasi yang belum lengkap.


```python
df = pd.read_csv('data_kantin.csv')

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

### Tugas Analisis 4

**Pertanyaan:**
Dari hasil df.info(), kolom mana yang jumlah non-null-nya lebih sedikit dari jumlah baris total? Apa artinya?

**Jawaban:**
Dari hasil df.info(), kolom menu dan terjual memiliki jumlah data non-null yang lebih sedikit dari jumlah baris total, yaitu 110 baris. Kolom menu memiliki 107 data non-null sehingga terdapat 3 data kosong, sedangkan kolom terjual memiliki 104 data non-null sehingga terdapat 6 data kosong. Artinya, kedua kolom tersebut memiliki missing value atau data kosong.

```python
print(df.isnull().sum())

df['terjual'] = df['terjual'].fillna(0)

df = df.dropna(subset=['menu'])
```

### Tugas Analisis 4

**Pertanyaan:**
Mengapa pada contoh di atas kolom terjual diisi (fillna) sedangkan baris dengan menu kosong justru dihapus (dropna)? Diskusikan alasan logisnya.

**Jawaban:**
Pada hasil pemeriksaan data, terdapat 6 data kosong pada kolom terjual dan 3 data kosong pada kolom menu. Kolom terjual diisi dengan 0 menggunakan fillna karena data kosong pada jumlah penjualan dapat dianggap sebagai tidak ada jumlah penjualan yang tercatat. Sedangkan baris dengan menu kosong dihapus menggunakan dropna karena nama menu merupakan informasi penting untuk melakukan analisis berdasarkan menu. Jika nama menu kosong, data tersebut sulit digunakan untuk mengetahui produk yang dianalisis.

```python
print(df.duplicated().sum())

df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)

print(df.dtypes)
```

### Tugas Analisis 5

**Pertanyaan:**
Setelah drop_duplicates(), berapa jumlah baris dataset sekarang dibanding sebelumnya? Mengapa penting memastikan tipe data (dtypes) sudah benar sebelum data dianalisis lebih lanjut?

**Jawaban:**
Sebelum dan setelah drop_duplicates(), jumlah baris dataset tetap 5 baris karena tidak terdapat data duplikat. Memastikan tipe data (dtypes) sudah benar penting agar setiap kolom dapat digunakan sesuai fungsinya dalam proses analisis. Contohnya, kolom harga bertipe integer sehingga dapat digunakan dalam perhitungan.

```python
laris = df[df['terjual'] > 20]

urut = df.sort_values(by='terjual', ascending=False)

df['total_pendapatan'] = df['harga'] * df['terjual']

ringkasan = df.groupby('menu')['total_pendapatan'].sum()

print(ringkasan)
```

### Tugas Analisis 6

**Pertanyaan:**
Dari hasil groupby di atas, menu apa yang menghasilkan total_pendapatan tertinggi? Bagaimana informasi ini bisa membantu pengambilan keputusan di kantin sekolah?

**Jawaban:**
Berdasarkan hasil groupby, menu yang menghasilkan total_pendapatan tertinggi adalah Nasi Goreng, yaitu sebesar Rp3.924.000.

Informasi ini dapat membantu kantin sekolah mengetahui menu yang menghasilkan pendapatan paling besar sehingga dapat menjadi bahan pertimbangan dalam mengatur penjualan dan persediaan menu.
